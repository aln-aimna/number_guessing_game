# Number Guessing Game

```python
#   Import Modules to use random and imported images
import random 
from IPython.display import Image
from IPython.display import clear_output

# Your loop or logic here
clear_output(wait=True) 
```

---

```python
# Function to get a valid guess within the difficulty range.
def input_num(max_number):
    while True:
        try:
            # Ask user to guess a number depends on difficulty or max_number
            guess = int(input(f"Guess a number (1-{max_number}): "))

            # Accept only number guesses within the valid game range.
            if 1 <= guess and guess <= max_number:
                return guess

            print(f"Please enter a number between 1 and {max_number}.")

        except ValueError:
            print("Please enter a valid number.")
```

```python
# Function to select the difficulty and return its max number.

def select_difficulty():
    while True:
        print("\nSelect Difficulty (1, 2, 3):")
        print("1 - Easy   (1-10)")
        print("2 - Normal (1-50)")
        print("3 - Hard   (1-100)")

        choice = input("Choose difficulty: ")

        if choice == "1":
            print("Game mode is Easy")
            return 10
        elif choice == "2":
            print("Game mode is Normal")
            return 50
        elif choice == "3":
            print("Game mode is Hard")
            return 100
        else:
            # Keep asking until the user selects a valid difficulty.
            print("Please choose 1, 2, or 3.")
```

```python
# Function - generate a random number based on the difficulty level
def random_num(max_number):
    
    # Generate a random number within the range set by the difficulty.
    return random.randint(1, max_number)

    # Use a fixed number for testing.
    # return 5
```

```python
# Function to guess a number 
def guess_number(max_number):

    # Get number from random_num function
    correct_num = random_num(max_number)

    while True:
        guess_num = input_num(max_number)

        # Clear the cell output
        clear_output(wait=True)

        if guess_num == correct_num:
            # print("Correct!")
            display(Image(filename="congrats.gif"))
            return

        elif guess_num > correct_num:
            # print("Lower")
            display(Image(filename="lower.gif"))

        else:
            # print("Higher")
            display(Image(filename="higher.gif"))

```

```python
# Function to start or play the game
def start_game():
    while True:
        # Ask the user for difficulty 
        max_number = select_difficulty()

        # Start the guessing number using the selected diffulty.
        guess_number(max_number)

        #Ask user to play again
        play_again = input("\nPlay again? (y/n): ").lower()

        clear_output(wait=True)

        # Exit loop when the player chooses no or "n"
        if play_again != "y":
            print("Thanks for playing!")
            break
```

```python
start_game()
```
